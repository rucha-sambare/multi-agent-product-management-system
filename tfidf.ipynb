```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df=pd.read_csv("final_labeled_feedback.csv")
```

```python
tfidf = TfidfVectorizer(
    max_features=7000,
    ngram_range=(1,2),
    min_df=3,
    max_df=0.90,
    sublinear_tf=True,
    strip_accents="unicode"
)
```

```python
X_tfidf = tfidf.fit_transform(df["clean_text"])
```

```python
print(X_tfidf.shape)
```
