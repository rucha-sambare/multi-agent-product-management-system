```python
df = df[df["review_text"].notna()]

df = df[df["review_text"].str.strip() != ""]
```

```python
import re
stop_words = set(stopwords.words("english"))

def clean_text(text):

    text = str(text)

    # lowercase
    text = text.lower()

    # remove urls
    text = re.sub(r"http\S+", "", text)

    # remove special chars
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # remove stopwords
    words = text.split()

    words = [
        word for word in words
        if word not in stop_words
    ]

    return " ".join(words)
df["clean_text"]=df["review_text"].apply(clean_text)
```

```python
df["clean_text"].isnull().sum()
```

```python
df["review_length"] = (
    df["clean_text"]
    .str.len()
)

df["word_count"] = (
    df["clean_text"]
    .str.split()
    .str.len()
)

df["has_reply"] = (
    df["reply_text"]
    .notna()
    .astype(int)
)

df["source"] = "playstore"
```
